```python
# Nombre y apellidos:
# [Escribe aquí tu nombre completo]

# Fecha:
# [Escribe la fecha de hoy, por ejemplo: 06/11/2025]

# Propósito del experimento:
# Entrenar y analizar un perceptrón multicapa (MLP) simple para comprender su estructura básica y comportamiento.

# --- Configuración del entorno ---
import tensorflow as tf
import numpy as np
import pandas as pd
import time

print("TensorFlow:", tf.__version__)
print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
print("Fecha y hora de ejecución:", time.strftime("%Y-%m-%d %H:%M:%S"))
```

```python
# --- Carga del dataset MNIST ---

from tensorflow import keras

# Cargar datos de entrenamiento y prueba
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()

# Verificar las dimensiones de los conjuntos
X_train.shape, X_test.shape
```

```python
# --- Reducción del dataset a un tamaño manejable ---

X_train, y_train = X_train[:10000], y_train[:10000]
X_test, y_test = X_test[:2000], y_test[:2000]

# Comprobamos el nuevo tamaño
X_train.shape, X_test.shape
```

```python
# --- Normalización y aplanado de las imágenes ---

# Escalar los valores de píxeles a rango [0, 1]
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Aplanar las imágenes de 28x28 → 784 píxeles por muestra
X_train = X_train.reshape((-1, 28 * 28))
X_test = X_test.reshape((-1, 28 * 28))

# Comprobamos las nuevas dimensiones
X_train.shape, X_test.shape
```

```python
# --- Definición del modelo MLP ---

from tensorflow.keras import layers
from tensorflow import keras

# Definición secuencial del modelo
model = keras.Sequential([
    layers.Input(shape=(784,)),           # Capa de entrada: 784 neuronas (una por píxel)
    layers.Dense(64, activation="relu"),  # Capa oculta: 64 neuronas, activación ReLU
    layers.Dense(10, activation="softmax")# Capa de salida: 10 neuronas (una por dígito 0–9)
])

# Resumen del modelo
model.summary()
```

```python
# --- Compilación del modelo MLP ---

model.compile(
    optimizer="adam",                            # Optimizador: ajusta los pesos automáticamente
    loss="sparse_categorical_crossentropy",      # Función de pérdida: mide el error de clasificación
    metrics=["accuracy"]                         # Métrica: porcentaje de aciertos
)
```

```python
# --- Entrenamiento del modelo MLP ---

history = model.fit(
    X_train, y_train,
    validation_split=0.1,   # 10% de los datos se usan para validación
    epochs=5,               # Número de veces que el modelo recorre todo el dataset
    batch_size=128,         # Número de muestras procesadas antes de actualizar los pesos
    verbose=1               # Muestra el progreso del entrenamiento
)
```

```python
# --- Evaluación del modelo en el conjunto de prueba ---

test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)

print("Pérdida en test:", round(test_loss, 4))
print("Precisión en test:", round(test_acc, 4))
```

```python
# --- Guardar el modelo entrenado ---
model.save("/content/mlp_model.h5")
print("Modelo guardado en /content/mlp_model.h5")
```

```python
# --- Generar informe en formato Markdown ---

summary = []
model.summary(print_fn=lambda x: summary.append(x))
summary = "\n".join(summary)

report = f"""
# Informe del experimento MLP

## Arquitectura
{summary}

## Métricas finales
- Test loss: {test_loss}
- Test accuracy: {test_acc}

## Fecha
{time.ctime()}
"""

# Guardar el informe en /content/
with open("/content/report.md", "w") as f:
    f.write(report)

# Mostrar el contenido generado
print(report)
```
